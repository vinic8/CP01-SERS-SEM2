```python
import pandas as pd
from google.colab import files
```

```python

# EXERCÍCIO 3 - POWER CONSUMPTION OF TETOUAN CITY

arquivo = "/content/Tetuan City power consumption (1).csv"
dados = pd.read_csv(arquivo)

print("Arquivo carregado:", arquivo)
print("Dimensões:", dados.shape)

dados = dados.rename(columns={
    "Zone 1 Power Consumption": "Consumo_Zona_1",
    "Zone 2  Power Consumption": "Consumo_Zona_2",
    "Zone 3  Power Consumption": "Consumo_Zona_3"
})


print("\n========== DADOS INICIAIS ==========")
display(dados.head())

max_zona_1 = dados["Consumo_Zona_1"].max()
max_zona_2 = dados["Consumo_Zona_2"].max()
max_zona_3 = dados["Consumo_Zona_3"].max()

print("\n========== CONSUMO MÁXIMO ==========")
print(f"Zona 1: {max_zona_1:.2f}")
print(f"Zona 2: {max_zona_2:.2f}")
print(f"Zona 3: {max_zona_3:.2f}")

maximos = {
    "Zona 1": max_zona_1,
    "Zona 2": max_zona_2,
    "Zona 3": max_zona_3
}

zona_maior = max(maximos, key=maximos.get)

coluna_maior = {
    "Zona 1": "Consumo_Zona_1",
    "Zona 2": "Consumo_Zona_2",
    "Zona 3": "Consumo_Zona_3"
}[zona_maior]

maximo = maximos[zona_maior]

print("\n========== MAIOR PICO ==========")
print(f"Zona com maior pico: {zona_maior}")
print(f"Maior consumo registrado: {maximo:.2f}")

limiar_70 = 0.70 * maximo

df_alto_consumo = dados[
    dados[coluna_maior] > limiar_70
].copy()

quantidade_alto = len(df_alto_consumo)
percentual_alto = quantidade_alto / len(dados) * 100

print("\n========== CONSUMO ACIMA DE 70% ==========")
print(f"Limiar de 70%: {limiar_70:.2f}")
print(f"Quantidade de registros: {quantidade_alto}")
print(f"Percentual da amostra: {percentual_alto:.2f}%")


temperatura_media = dados["Temperature"].mean()

print("\n========== TEMPERATURA ==========")
print(f"Temperatura média: {temperatura_media:.2f}")


df_alto_temperatura = dados[
    (dados[coluna_maior] > limiar_70) &
    (dados["Temperature"] > temperatura_media)
].copy()

quantidade_com_temperatura = len(df_alto_temperatura)
percentual_com_temperatura = (
    quantidade_com_temperatura / len(dados) * 100
)

print("\n========== CONSUMO ALTO + TEMPERATURA ALTA ==========")
print(f"Quantidade de registros: {quantidade_com_temperatura}")
print(f"Percentual da amostra: {percentual_com_temperatura:.2f}%")


reducao = quantidade_alto - quantidade_com_temperatura

if quantidade_alto > 0:
    percentual_reducao = reducao / quantidade_alto * 100
else:
    percentual_reducao = 0

print("\n========== COMPARAÇÃO ==========")
print(f"Somente consumo acima de 70%: {quantidade_alto}")
print(f"Consumo acima de 70% + temperatura acima da média: {quantidade_com_temperatura}")
print(f"Registros eliminados: {reducao}")
print(f"Redução: {percentual_reducao:.2f}%")

print("\n========== RESPOSTAS ==========")

print(f"""
1. A zona que apresentou o maior pico de consumo foi a {zona_maior},
com um valor máximo de {maximo:.2f}.

2. O limite de 70% foi utilizado para identificar períodos de consumo
elevado em relação ao maior consumo registrado na zona analisada.

3. Após adicionar a temperatura como segundo critério, a quantidade
de registros passou de {quantidade_alto} para {quantidade_com_temperatura}.
Isso ocorre porque agora o registro precisa apresentar simultaneamente
consumo acima de 70% do máximo e temperatura acima da média.

4. A análise permite identificar períodos de maior demanda elétrica e
observar se esses períodos também ocorreram quando a temperatura estava
acima da média. Isso ajuda a analisar uma possível relação entre as
condições ambientais e o consumo de energia.
""")
```
