```python
import requests
import pandas as pd

# 0. Garante que os dados da API existam na memória
url = "https://apicarga.ons.org.br/prd/cargaverificada"
parametros = {
    "dat_inicio": "2025-08-01",
    "dat_fim": "2025-08-07",
    "cod_areacarga": "SP"
}

response = requests.get(url, params=parametros, timeout=30)
dados_json = response.json()

if isinstance(dados_json, list):
    registros = dados_json
elif isinstance(dados_json, dict):
    chaves_com_lista = [k for k, v in dados_json.items() if isinstance(v, list)]
    registros = dados_json[chaves_com_lista[0]]

# 1. Cria o DataFrame base
df_base = pd.DataFrame(registros)

# 2. Renomeia as colunas
mapeamento_colunas = {
    "dat_inspecao": "data_hora",
    "din_instante": "data_hora",
    "val_carga": "carga",
    "val_cargaverificada": "carga",
    "nom_areacarga": "area",
    "cod_areacarga": "area",
}
dados_tratados = df_base.rename(columns=mapeamento_colunas)

# 3. Filtra colunas relevantes
colunas_desejadas = [c for c in ["data_hora", "area", "carga"] if c in dados_tratados.columns]
dados_tratados = dados_tratados[colunas_desejadas].copy()

# 4. Verifica valores ausentes
valores_nulos = dados_tratados.isnull().sum()

# 5. Converte 'carga' para numérico
if "carga" in dados_tratados.columns:
    dados_tratados["carga"] = pd.to_numeric(dados_tratados["carga"], errors="coerce")

# 6. Converte 'data_hora' para datetime
if "data_hora" in dados_tratados.columns:
    dados_tratados["data_hora"] = pd.to_datetime(dados_tratados["data_hora"], errors="coerce")

# Define a variável global 'dados' para os próximos desafios
dados = dados_tratados

# Saída
print("--- DESAFIO 2 CONCLUÍDO COM SUCESSO ---")
print("\nPrimeiros registros:")
print(dados.head())

print("\nValores nulos por coluna:")
print(valores_nulos)

print("\nTipos de dados:")
print(dados.dtypes)
```
